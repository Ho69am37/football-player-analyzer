```python
match_id = 22912

with open(f"../data/events/{match_id}.json", "r", encoding="utf-8") as f:
    events = json.load(f)

print("Total events: ", len(events))
print("\nfirst Event: ", events[0])
```

```python
print("=== First event (full structure) ===")
print(json.dumps(events[0], indent=2, ensure_ascii=False))
```

```python

event_types = {}
for e in events:
    t = e['type']['name']
    event_types[t] = event_types.get(t, 0) + 1

print("=== Event types ===")
for t, count in sorted(event_types.items(), key=lambda x: -x[1]):
    print(f"{t}: {count}")
```

```python
pass_events = [e for e in events if e['type']['name'] == 'Pass']
print("Total passes:", len(pass_events))
print(pass_events[0])
```

```python
players = set()

for e in events:
    if 'player' in e:
        players.add(e['player']['name'])

print(f"total of players : {len(players)}")
print("\nAll players: ")
print(sorted(players))
```

```python
print(f"Total players: {len(players)}")
print("\n=== All players ===")
for i, p in enumerate(sorted(players), 1):
    print(f"{i:2d}. {p}")
```

```python
def get_player_stats(events, player_name):
    player_events = [e for e in events if e.get('player', {}).get('name') == player_name]
    stats = {
        'total_events': len(player_events),
        'passes': 0,
        'passes_completed': 0,
        'shots': 0,
        'goals': 0,
        'carries': 0,
        'pressures': 0,
        'duels': 0,
    }

    for e in player_events:
        t = e['type']['name']

        if t == 'Pass':
            stats['passes'] += 1
            if 'outcome' not in e.get('pass', {}):
                stats['passes_completed'] += 1
        elif t == 'Shot':
            stats['shots'] += 1
            if e.get('shot', {}).get('outcome', {}).get('name', {}) == 'Goal':
                stats['goals'] += 1
        elif t == 'Carry':
            stats['carries'] += 1
        elif t == 'Pressure':
            stats['pressures'] += 1
        elif t == 'Duel':
            stats['duels'] += 1

    return stats

stats = get_player_stats(events, 'Mohamed Salah')
print("===Mohamed Salah===")
for k, v in stats.items():
    print(f" {k}: {v}")                            
```

```python
players_to_compare = [
    'Mohamed Salah',
    'Sadio Mané',
    'Roberto Firmino',   
    'Harry Kane',
    'Heung-Min Son',
    'Christian Dannemann Eriksen',
]

for p in players_to_compare:
    stats = get_player_stats(events, p)
    if stats['total_events'] == 0:
        print("\n {p} not found!")
        continue
    pass_pct = stats['passes'] / max(stats['passes_completed'], 1) * 100
    print(f"\n==={p}===")
    print(f" Events: {stats['total_events']}")
    print(f"  Passes: {stats['passes_completed']}/{stats['passes']} ({pass_pct:.1f}%)")
    print(f"  Shots: {stats['shots']} | Goals: {stats['goals']}")
    print(f"  Carries: {stats['carries']}")
    print(f"  Pressures: {stats['pressures']}")
```
