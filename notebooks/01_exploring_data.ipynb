```python
import json 
import pandas as pd 

with open("../data/competitions.json", "r", encoding = "utf-8") as f:
    competitions = json.load(f)

print(type(competitions))
print(len(competitions))
print(competitions[0])
```

```python
df = pd.DataFrame(competitions)

print("Shape: ", df.shape)
print("\nColumns: ", df.columns.tolist())
print("\nfirst 5 raws: ")
df.head()
```

```python
print("Unique Competitions: ", df["competition_name"].nunique())
print("\nAll competitions: ")
print(df["competition_name"].unique())
```

```python
cl =  df[df["competition_name"] == 'Champions League']
print("champions league seasons: ", len(cl))
print("\nAll Champions League seasons:")
print(cl[['competition_id', 'season_id', 'season_name']])
```

```python
import json

with open("../data/matches/16/4.json", "r", encoding="utf-8") as f:
    matches = json.load(f)

print("Type : ", type(matches))
print("\nNumber of matches : ", len(matches))
for m in matches:
    home = m['home_team']['home_team_name']
    away = m['away_team']['away_team_name']
    print(f" {m['match_date']} | {home} vs {away} | match_id: {m['match_id']}")
```

```python
match_id = 22912

with open(f"../data/events/{match_id}.json", "r", encoding="utf-8") as f:
    events = json.load(f)

print("Total events: ", len(events))
print("\nfirst Event: ", events[0])
```

```python
print("=== First event (full structure) ===")
print(json.dumps(events[0], indent=2, ensure_ascii=False))
```

```python

event_types = {}
for e in events:
    t = e['type']['name']
    event_types[t] = event_types.get(t, 0) + 1

print("=== Event types ===")
for t, count in sorted(event_types.items(), key=lambda x: -x[1]):
    print(f"{t}: {count}")
```
